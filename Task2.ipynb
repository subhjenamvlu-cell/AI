2. Implement Breadth-First Search to find the shortest path in a simple maze. 

```python
from collections import deque

def bfs(maze, start, goal):
    rows, cols = len(maze), len(maze[0])
    queue = deque([(start, [start])])
    visited = set([start])

    while queue:
        (x, y), path = queue.popleft()

        if (x, y) == goal:
            return path

        # Possible moves: Right, Down, Left, Up
        for dx, dy in [(0,1),(1,0),(0,-1),(-1,0)]:
            nx, ny = x + dx, y + dy

            if 0 <= nx < rows and 0 <= ny < cols and maze[nx][ny] == 0:
                if (nx, ny) not in visited:
                    visited.add((nx, ny))
                    queue.append(((nx, ny), path + [(nx, ny)]))

    return None

# Example Maze (0 = path, 1 = wall)
maze = [
    [0, 0, 1],
    [0, 0, 0],
    [1, 0, 0]
]

start = (0, 0)
goal = (2, 2)

path = bfs(maze, start, goal)
print("Shortest Path:", path)
```

3. Apply Bayes' Rule to calculate the probability of rain given cloudy weather. 


```python
# Given probabilities
P_rain = 0.3
P_cloudy_given_rain = 0.8
P_cloudy = 0.5

# Bayes' Theorem
P_rain_given_cloudy = (P_cloudy_given_rain * P_rain) / P_cloudy

# Output
print("Probability of Rain given Cloudy:", P_rain_given_cloudy)
```

4. Solve the Missionaries and Cannibals problem using BFS or DFS. 

```python
#BFS
from collections import deque

# Check if state is valid
def is_valid(m, c):
    return (m == 0 or m >= c) and (3 - m == 0 or (3 - m) >= (3 - c))

def bfs():
    start = (3, 3, 1)  # (missionaries, cannibals, boat)
    goal = (0, 0, 0)

    queue = deque([(start, [])])
    visited = set()

    while queue:
        (m, c, boat), path = queue.popleft()

        if (m, c, boat) == goal:
            return path + [(m, c, boat)]

        # Possible moves
        moves = [(1,0),(2,0),(0,1),(0,2),(1,1)]

        for dm, dc in moves:
            if boat == 1:  # boat on left
                new = (m - dm, c - dc, 0)
            else:          # boat on right
                new = (m + dm, c + dc, 1)

            if 0 <= new[0] <= 3 and 0 <= new[1] <= 3 and is_valid(new[0], new[1]):
                if new not in visited:
                    visited.add(new)
                    queue.append((new, path + [(m, c, boat)]))

solution = bfs()
print("Solution Path:", solution)
```

5. Implement A* Search for a graph and compare its result with UCS.

```python
import heapq

# Graph representation (adjacency list)
graph = {
    'A': [('B', 1), ('C', 3)],
    'B': [('D', 1), ('E', 5)],
    'C': [('F', 2)],
    'D': [],
    'E': [('F', 1)],
    'F': []
}

# Heuristic values (estimated cost to goal F)
heuristic = {
    'A': 6,
    'B': 4,
    'C': 2,
    'D': 4,
    'E': 1,
    'F': 0
}

# ---------------------------
# A* Search
# ---------------------------
def astar(start, goal):
    pq = [(0 + heuristic[start], 0, start, [])]  # (f, g, node, path)
    visited = set()

    while pq:
        f, g, node, path = heapq.heappop(pq)

        if node in visited:
            continue

        path = path + [node]
        visited.add(node)

        if node == goal:
            return path, g

        for neighbor, cost in graph[node]:
            heapq.heappush(pq, (g + cost + heuristic[neighbor], g + cost, neighbor, path))

    return None


# ---------------------------
# Uniform Cost Search (UCS)
# ---------------------------
def ucs(start, goal):
    pq = [(0, start, [])]  # (cost, node, path)
    visited = set()

    while pq:
        cost, node, path = heapq.heappop(pq)

        if node in visited:
            continue

        path = path + [node]
        visited.add(node)

        if node == goal:
            return path, cost

        for neighbor, c in graph[node]:
            heapq.heappush(pq, (cost + c, neighbor, path))

    return None


# Run both algorithms
print("A* Result:", astar('A', 'F'))
print("UCS Result:", ucs('A', 'F'))
```
